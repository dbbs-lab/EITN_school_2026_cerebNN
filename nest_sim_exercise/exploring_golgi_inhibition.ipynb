# Exploring Golgi Inhibition

Building on the previous BSB–NEST simulation tutorial, we will explore the effects of connectivity changes on network activity.
In particular, we will study the role of Golgi cells.

schema_grgo.png

Golgi cells receive:


*   excitatory input from mossy fibers
*   excitatory input from granule cells
*   inhibition input from other Golgi cells

and inhibit granule cells.

To study the role of these connections we could:


*   remove mf-Goc connections (no feedforward excitation condition, noFF)
*   remove Grc-GoC connections (no feedback excitation condition, noFB)
*   remove GoC-GrC connections (no inhibition at all, noGABA)

**Exercise**: Compare the canonical circuit against the noFF, noFB, and noGABA conditions under a predefined *stimulus protocol*.

*Stimulus protocol*: $N$ active mossy fibers receive an input at a frequency of $f_{MF}$ Hz, generated according to a specific distribution (e.g., a homogeneous Poisson process or simultaneous input), within a time window of $t_{stim}$ ms starting at $t_{start}$ ms. Both active and inactive mossy fibers may or may not receive background input.
