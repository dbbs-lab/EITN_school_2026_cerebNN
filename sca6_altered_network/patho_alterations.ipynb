# SCA6 ataxia-like cerebellar cortical network

**SCA6 ataxia**:


*   Purkinjie cells show reduced firing rates
*   they also show morphologocal alterations resulting in reduced convergences of both excitatory GrC input and inhibitory MLI input.

2.   *network-level alteration*: due to the morphological atrophy, convergence from granule cells both ascending axon and parallel fiber, is reduced by 24%, and convergence from MLIs through both stellate and basket cells, is reduces by 8%
