# SCA6 ataxia-like cerebellar cortical network

Compare control and pathological dynamical properties by simulating a set of stimulus protocols designed to highlight the effects of the intrinsically lower PC excitability in the SCA6 ataxia model.

*   background Poisson noise to all mossy fibers
*   background Poisson noise to all + localized input
*   control vs. pathoogical input/output frequency relationship
*   ...
